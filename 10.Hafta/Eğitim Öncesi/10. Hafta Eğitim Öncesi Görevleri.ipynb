# Problem 2

```python
import numpy as np

# 1 - 0 ile 10 arasında bulunan tek sayılardan oluşan matris üretin
tek_sayilar = np.arange(1, 11, 2)  # [1, 3, 5, 7, 9]
print("Tek sayılardan oluşan matris:", tek_sayilar)

# 2 - Oluşturduğunuz matrisi parçalayın
parca1, parca2 = np.split(tek_sayilar, 2)  # Matrisi iki parçaya bölüyoruz
print("Birinci parça:", parca1)
print("İkinci parça:", parca2)

# 3 - Parçalanmış matrisleri birleştirin
birlesmis_matris = np.concatenate((parca1, parca2))
print("Birleştirilmiş matris:", birlesmis_matris)

```
