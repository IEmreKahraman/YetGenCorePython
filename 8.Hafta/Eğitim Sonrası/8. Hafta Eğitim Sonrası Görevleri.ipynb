# Problem 1

Şimdi beraber kullanıcılardan aldığımız verileri işleyerek, onlara hem çıktı verecek, hem de aldığımız ve hesapladığımız verileri ileride kullanabilmek adına kaydedeceğiz.


 **Bu bir veri giriş platformu olacak**


 * kullanıcıya kaç giriş yapacağını soracağız
* kullanıcıdan isim-soyisim, yaş, boy, kilo bilgilerini alarak VKİ hesaplayacağız
* ardından bunları sınıflara göre kaydedeceğiz bir dosyaya kaydedeceğiz


Sınıflar ise şu şekilde: 
* 18, 5 kg/m.'nin altında olanlar: Zayıf
* 18.5 – 24, 9 kg/m. arasında olanlar: Normal kilolu
* 25 – 29, 9 kg/m. arasında olanlar: Fazla kilolu
* 30 – 39, 9 kg/m. arasında olanlar: Obez
* 40 kg/m.'nin üzerinde olanlar: İleri derecede obez (morbid obez), olarak görülür.

```python
def vki_hesapla(kilo, boy):
    return kilo / (boy ** 2)

def vki_sinifi(vki):
    if vki < 18.5:
        return "Zayıf"
    elif 18.5 <= vki < 25:
        return "Normal kilolu"
    elif 25 <= vki < 30:
        return "Fazla kilolu"
    elif 30 <= vki < 40:
        return "Obez"
    else:
        return "İleri derecede obez (morbid obez)"

def verileri_kaydet(dosya_adi, veriler):
    with open(dosya_adi, "a") as dosya:
        for veri in veriler:
            dosya.write(f"{veri['isim']}, {veri['yas']}, {veri['boy']}, {veri['kilo']}, {veri['vki']:.2f}, {veri['vki_sinifi']}\n")

def veri_giris_platformu():
    giris_sayisi = int(input("Kaç giriş yapacaksınız? "))
    veriler = []

    for _ in range(giris_sayisi):
        isim = input("İsim-Soyisim: ")
        yas = int(input("Yaş: "))
        boy = float(input("Boy (metre cinsinden): "))
        kilo = float(input("Kilo (kg cinsinden): "))

        vki = vki_hesapla(kilo, boy)
        vki_sinifi_ = vki_sinifi(vki)

        veri = {
            "isim": isim,
            "yas": yas,
            "boy": boy,
            "kilo": kilo,
            "vki": vki,
            "vki_sinifi": vki_sinifi_
        }

        veriler.append(veri)
        print(f"{isim}, VKİ: {vki:.2f}, Sınıf: {vki_sinifi_}")

    verileri_kaydet("kullanici_verileri.txt", veriler)
    print("Tüm veriler kaydedildi.")

veri_giris_platformu()
```

# Problem 2
**YetBar - Smoothie Bank**
---
YetBar spor salonunun girişinde bulunan, insansız bir sağlıklı içecek otomatıdır. Kullanıcı çeşitli girdiler smoothie içeceğine koyacağı ürünleri belirleyebilir, fiyat hesaplaması yaptırabilir ya da sadece menüyü görüntüleyebilir.

**Projenin Amacı:**

*   Birbirini etkileyen ve bir kod içerisinde birden fazla defa başvurulacak fonksiyonların oluşturulup, "fonksiyonel" proglamlamak.
*   Büyük programları yazarken, yazılımcı olarak yapabileceğimiz hatalardan kaçınmak için kontrol yöntemleri geliştirmek.
*   Kullanıcı inputlarını ve yaptığımız hesapları, raporlamak/loglamak amacıyla kaydetmek.

**Programdan Beklenen Çıktılar**
* Program kullanıcıya temelde 2 farklı işlem yapma imkanı sağlar.
> 1. Menüyü görüntüleme
> 3. Satın alma
* Satın alınan her içecek tarih, fiyat bilgisi ile kaydedilmeli ve her zaman ulaşılabilir olmalıdır.




```python
import datetime

menu = {
    "Muz": 2.5,
    "Çilek": 3.0,
    "Ananas": 4.0,
    "Kivi": 3.5,
    "Elma": 2.0,
    "Ispanak": 1.5,
    "Yulaf": 2.5,
    "Badem Sütü": 3.5,
    "Hindistan Cevizi Suyu": 4.5
}


def menu_goruntule():
    print("\nYetBar Smoothie Menüsü:")
    for urun, fiyat in menu.items():
        print(f"{urun}: {fiyat} TL")
    print()


def satin_al():
    secilen_urunler = []
    toplam_fiyat = 0.0

    while True:
        urun = input("Eklemek istediğiniz ürünü girin (veya 'bitir' yazarak işlemi sonlandırın): ")
        
        if urun.lower() == "bitir":
            break
        elif urun in menu:
            secilen_urunler.append(urun)
            toplam_fiyat += menu[urun]
            print(f"{urun} eklendi. Mevcut toplam fiyat: {toplam_fiyat} TL")
        else:
            print("Menüde böyle bir ürün yok, lütfen geçerli bir ürün girin.")

    if secilen_urunler:
        print(f"\nSatın aldığınız ürünler: {', '.join(secilen_urunler)}")
        print(f"Toplam fiyat: {toplam_fiyat} TL")
        satin_alma_kaydet(secilen_urunler, toplam_fiyat)
    else:
        print("Herhangi bir ürün seçilmedi.")


def satin_alma_kaydet(urunler, fiyat):
    tarih = datetime.datetime.now().strftime("%Y-%m-%d %H:%M:%S")
    with open("satin_alma_log.txt", "a") as dosya:
        dosya.write(f"Tarih: {tarih}, Ürünler: {', '.join(urunler)}, Toplam Fiyat: {fiyat} TL\n")
    print("Satın alma işlemi kaydedildi.\n")

def yetbar():
    while True:
        print("YetBar'a Hoşgeldiniz!")
        print("1. Menüyü görüntüle")
        print("2. Satın alma işlemi")
        print("3. Çıkış yap")

        secim = input("Bir seçenek girin: ")

        if secim == "1":
            menu_goruntule()
        elif secim == "2":
            satin_al()
        elif secim == "3":
            print("YetBar'dan çıkış yapılıyor. İyi günler!")
            break
        else:
            print("Geçersiz seçim, lütfen tekrar deneyin.")
yetbar()
```
